used for skewed distribution

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('placements.csv')
df.sample(5)
```

```python
sns.histplot(df['placement_exam_marks'],kde=True)
plt.show()
```

As its right skewed, we can use IQR method

```python
Q1=df['placement_exam_marks'].quantile(0.25)
Q3=df['placement_exam_marks'].quantile(0.75)
Q1,Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
upper=Q3+1.5*IQR
lower=Q1-1.5*IQR
upper,lower
```

Now either trim or cap the values according to upper and lower values
