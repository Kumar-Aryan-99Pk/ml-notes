```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_regression
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg=LinearRegression()
reg.fit(X,y)
```

```python
reg.coef_ , reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='purple')
```

<h2>Performing Gradient Descent ,taking m const

<b>Lets apply GD assuming slope is const m=78.35 <br>
<b>Taking b=0 initially

```python
y_pred=((78.35*X)+0).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='purple',label='OLS')
plt.plot(X,y_pred,color='red',label='b=0')
plt.legend()
plt.show()
```

```python
m=78.35
b=0

loss_slope=-2*np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
lr=0.1 #eta
step_size=lr*loss_slope
step_size
```

```python
b=b-step_size
b
```

```python
y_pred1=((78.35*X)+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='purple',label='OLS')
plt.plot(X,y_pred,color='gray',label='b=0')
plt.plot(X,y_pred1,color='red',label='b={}'.format(b))
plt.legend()
plt.show()
```

```python
loss_slope=-2*np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
step_size=lr*loss_slope
step_size
```

```python
b=b-step_size
b
```

```python
y_pred2=((78.35*X)+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='purple',label='OLS')
plt.plot(X,y_pred,color='gray',label='b=0')
plt.plot(X,y_pred1,color='gray',label='b=20.92')
plt.plot(X,y_pred2,color='red',label='b={}'.format(b))
plt.legend()
plt.show()
```

```python
loss_slope=-2*np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
step_size=lr*loss_slope
step_size
```

```python
b=b-step_size
b
```

```python
y_pred3=((78.35*X)+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='purple',label='OLS')
plt.plot(X,y_pred,color='gray',label='b=0')
plt.plot(X,y_pred1,color='gray',label='b=20.92')
plt.plot(X,y_pred2,color='gray',label='b=25.11')
plt.plot(X,y_pred3,color='red',label='b={}'.format(b))
plt.legend()
plt.show()
```

```python
b=0
m=78.35
lr=0.1

epochs=10

for i in range(epochs):
    loss_slope=-2*np.sum(y-m*X.ravel()-b)
    b=b-(lr*loss_slope)

    y_pred=m*X+b
    plt.plot(X,y_pred)

plt.scatter(X,y)
```

<h2>Performing Gradient descent by adding m

```python
m=10
b=20
loss_slope_b=-2*np.sum(y-m*X.ravel()-b)
loss_slope_m=-2*np.sum((y-m*X.ravel()-b)*X.ravel())

m=m-(lr*loss_slope_m)
b=b-(lr*loss_slope_b)
m,b
```
