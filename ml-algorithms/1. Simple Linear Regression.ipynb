```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('datasets/package.csv')
df.head()
```

```python
sns.scatterplot(x=df['cgpa'],y=df['package'])
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
X=df.iloc[:,0:1]
y=df.iloc[:,1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
X.shape,y.shape
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
lr.coef_,lr.intercept_
```

```python
sns.lineplot(x=X_train.values.reshape(-1),y=lr.predict(X_train),color='red')
sns.scatterplot(x=df['cgpa'],y=df['package'])
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
mean_absolute_error(y_test,y_pred),mean_squared_error(y_test,y_pred)
```
