## Working with dates

```python
date.info()
```

```python
date['date']=pd.to_datetime(date['date'])
date.info()
```

```python
#Extract year
date['date_year']=date['date'].dt.year
date.sample(5)
```

```python
#Extract month no.
date['month_no']=date['date'].dt.month
date.sample(5)
```

```python
#Extract month name
date['month_name']=date['date'].dt.month_name()
date.sample(5)
```

```python
#Extract date name
date['date_day']=date['date'].dt.day
date.sample(5)
```

```python
#Extract day of week
date['date_dow']=date['date'].dt.dayofweek
date.sample(5)
```

```python
#Extract dow name
date['date_dow_name']=date['date'].dt.day_name()
date.sample(5)
```

```python
#Extract week of year
date['date_week']=date['date'].dt.isocalendar().week
date.sample(5)
```

```python
#Extract quarter of year
date['quarter']=date['date'].dt.quarter
date.sample(5)
```

```python
#Etract time elapsed between dates

import datetime
today=datetime.datetime.today()
today
```

```python
today-date['date']
```

```python
(today-date['date']).dt.days
```
